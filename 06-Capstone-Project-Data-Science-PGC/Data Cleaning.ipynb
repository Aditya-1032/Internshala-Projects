```python
# Import the prerequisits
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'C:\Users\Aditya\OneDrive\Documents\Intershala Data\Pyhton\Deep Learning\Web Scrapping\books_full_dataset.csv')
df.head()
```

### To Handle missing values & duplicates 

```python
print("Missing values")
print(df.isnull().sum())
print("------------------")
print("Duplicate values")
print(df.duplicated().sum())
```

    The dataset was inspected for missing values using isnull(). Since the source website provides for all attributes on the listing pages, no missing values were observed.
    
    Duplicate book records were checked and no record for the duplicates were found.

### Fix inconsistent formats (date, text, numbers)

```python
# Check data type
df["Price"] = df["Price"].astype(float)
```

Price column was converted to numeric format for statistical analysis.

```python
# Numerical Validation
df['Rating'].unique()
```

```python
# Clean Text columns
df["Title"] = df["Title"].str.strip()
df["Genre"] = df["Genre"].str.strip()
df["Availability"] = df["Availability"].str.strip()
```

Text fields were standardized by removing extra whitespace to ensure uniform formatting.

### Feature Engineering

```python
# Rating Percentage
df['Rating_Percent'] = (df['Rating']/5)*100
```

Ratings were converted into percentage form to improve interpretability and comparison.

```python
# Price Band
df["Price_Band"] = pd.cut(
    df["Price"],
    bins=[0, 20, 40, 60],
    labels=["Low", "Medium", "High"]
)
```

Books were categorized into price bands to facilitate pricing analysis across genres. 

```python
# Availability Flag
df["In_Stock"] = df["Availability"].apply(lambda x: 1 if "In stock" in x else 0)
```

```python
# Title Length
df['Title_Length'] = df['Title'].apply(len)
```

Title length was engineered as a textual feature to explore potential relationships with price and ratings.

### Convert raw, messy data → structured analytics table 

```python
df['In_Stock'].unique()
```

    Variance = 0, The feature carrries no information.
    The model can not learn anything from it.

```python
# Drop raw/unnecessary column
df_cleaned = df.drop(columns = ['Availability', 'In_Stock'])
```

```python
# Reorder column
df_cleaned = df_cleaned[
    [
        "Title",
        "Genre",
        "Price",
        "Price_Band",
        "Rating",
        "Rating_Percent",
        "Title_Length"
    ]
]
```

```python
# Saving cleaned dataset
df_cleaned.to_csv('Books_cleaned_dataset.csv', index = False)
```

In Phase 2, the raw scraped data was cleaned by validating missing values, removing duplicate records, and standardizing numeric and text formats. Additional features such as rating percentage, price bands, stock indicators, and title length were engineered to enhance analytical depth. The final output is a structured, analysis-ready dataset.
